# Play with CHSM8

Ask for a style in plain words and see the move distribution change, or play a game against Stockfish.

```python
import os, sys
sys.path.insert(0, ".." if os.path.basename(os.getcwd()) == "notebooks" else ".")
from chsm8.play import Player
p = Player.from_pretrained("LegumMagister/chsm8")   # needs access to google/embeddinggemma-300m
```

```python
game = "1. e4 c5 2. Nf3 d6 3. d4 cxd4 4. Nxd4 Nf6 5. Nc3 a6"
for prompt in ["", "attack the king at any cost", "play safe, trade pieces, aim for a draw"]:
    top = p.scores(game, prompt)[:5]
    print(f"{prompt or '(no prompt)':45s}", "  ".join(f"{m['san']} {m['logp']:.2f}" for m in top))
```

Any notation works for the game: SAN, UCI, LAN or a PGN with move numbers. `move()` samples with a temperature.

```python
print(p.move("e2e4 e7e5 g1f3", "sacrifice material for the initiative", temperature=1.0))
```

## A game against Stockfish

Our evaluation setting: Stockfish 17.1 at UCI_Elo 2000, 0.05 s per move. Point `STOCKFISH` to the binary.

```python
from chsm8.play import play_stockfish
STOCKFISH = os.environ.get("STOCKFISH", "stockfish")
g = play_stockfish(p, STOCKFISH, elo=2000, model_white=True, prompt="attack the king at any cost")
print(g)
```

## Player models

Fine-tunes on one player's games, without text (chsm8-player-carlsen, -kasparov, ...).

```python
carlsen = Player.from_pretrained("LegumMagister/chsm8-player-carlsen")
print(carlsen.scores(game)[:5])
```
